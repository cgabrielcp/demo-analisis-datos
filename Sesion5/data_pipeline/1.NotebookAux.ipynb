```python
# El primer isnull te regresa si existe elementos nullos
customers_df[['customer_fname', 'customer_lname', 'customer_email']].isnull()
```

```python
# El primer any te asegurra que no si existe en cada columna un valor true
customers_df[['customer_fname', 'customer_lname', 'customer_email']].isnull().any()
```

```python
# El segundo any te asegurra que no si existe en cada una de los 3 elementos seleccionados
customers_df[['customer_fname', 'customer_lname', 'customer_email']].isnull().any().any()
```
